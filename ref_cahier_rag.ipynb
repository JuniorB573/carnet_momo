## 1. Contexte

La transition numérique au Bénin s'accélère, entraînant une explosion des usages mais aussi des litiges en ligne (cyberharcèlement, usurpation d'identité, fraudes). Les textes qui encadrent ces usages sont longs, techniques et inaccessibles pour le citoyen ordinaire. 

**Le problème** : Un citoyen ne parle pas le langage de la loi ("atteinte à l'intimité") mais le langage du quotidien ("ma photo a été publiée sur Facebook sans mon accord"). De plus, l'utilisation d'une Intelligence Artificielle générative classique pour répondre à ces questions est dangereuse : elle risque d'inventer des lois (hallucinations) ou de citer des textes étrangers.

**Notre solution** : Une approche **RAG (Retrieval-Augmented Generation) stricte et sur-mesure**. Le système agit comme un pont : il comprend le citoyen, retrouve l'article exact dans un corpus de lois, et traduit la loi en une réponse claire, sans jamais extrapoler.

## 2. Objectifs d'Impact

**Objectif général** : Démocratiser l'accès à la justice numérique au Bénin en permettant à toute personne d'obtenir une réponse claire, juridiquement exacte et instantanée.

**Objectifs spécifiques :**
- **Vulgarisation sans dénaturation** : Rendre le droit compréhensible sans aucune formation juridique.
- **Zéro Hallucination et Traçabilité totale** : Chaque réponse doit obligatoirement renvoyer à la loi et à l'article d'origine. Si le texte ne couvre pas la question, l'IA doit avoir l'intelligence de refuser de répondre.
- **Inclusion numérique** : Fournir une application web ultra-légère, pensée pour un usage mobile.

## 3. Périmètre d'Intervention

**Inclus :**
- Loi n° 2017-20 du 20 avril 2018 portant Code du numérique en République du Bénin.
- Loi n° 2018-16 portant Code pénal en République du Bénin.
- Traitement des questions en français sur la vie privée, les données personnelles et la cybercriminalité.

**Exclus :**
- Conseil juridique personnalisé ou aide à la constitution de dossiers (l'outil informe, il ne remplace pas un avocat).
- Textes non fournis dans le corpus initial.

## 4. Utilisateurs Cibles

- **Le Grand Public** : Citoyens et jeunes internautes vulnérables aux situations litigieuses en ligne.
- **Les Professionnels** : Journalistes, associations et étudiants cherchant à retrouver un article précis en quelques secondes.
- **Prérequis** : Aucune compétence technique ni juridique n'est nécessaire.

## 6. Exigences Non Fonctionnelles (*)

- **Fiabilité absolue** : Interdiction stricte de modifier les durées de peines ou les montants d'amendes.
- **Performance** : Pipeline optimisé pour un temps de réponse rapide, compatible grand public.
- **Privacy by Design** : Aucune conservation des données personnelles ; architecture respectueuse de l'anonymat.
- **Maintenabilité** : Code modulaire permettant l'ajout de nouvelles lois par simple dépôt de fichier.
- **Sobriété** : Architecture conçue pour tourner sur des services cloud à très bas coût.

## 7. Architecture Technique

Pour atteindre ce niveau de fiabilité, le projet s'appuie sur un pipeline RAG de pointe, bien supérieur aux solutions standards :

1. **Ingestion Intelligente** : Lecture et nettoyage des PDF institutionnels (souvent altérés ou mal formatés) avec des heuristiques de correction. Découpage logique par article.
2. **Indexation** : Création d'un espace vectoriel (embeddings multilingues `BAAI/bge-m3`) et d'un index lexical.
3. **Recherche Hybride (Fusion RRF)** : Croisement de la recherche sémantique (FAISS) et lexicale (BM25) pour une précision chirurgicale.
4. **Re-ranking (Cross-Encoder)** : Un second modèle d'IA classe les résultats et filtre les documents non pertinents avant même de les envoyer au LLM.
5. **Génération Sécurisée** : Appel au LLM (Google Gemini) avec un prompt le forçant à agir comme un synthétiseur strict, sans improvisation.
6. **Interface & Déploiement** : Application Streamlit propulsée depuis GitHub vers Streamlit Community Cloud.

## 8. Évaluation et Critères d'Acceptation

| Critère | Méthode de mesure | Objectif visé |
|---|---|---|
| **Précision du Moteur** | Mesures Hit@k et MRR sur un jeu de test annoté | Le bon article doit être dans le Top 3 |
| **Zéro Hallucination** | Relecture manuelle et tests de stress (questions pièges) | 100 % de refus sur les questions hors corpus |
| **Fidélité Juridique** | Vérification des montants/peines cités | 100 % d'exactitude des citations |
| **Robustesse Technique**| Tests de charge et de bascule (fallback) | Disponibilité continue de l'application |

## 11. Gestion des Risques

| Risque | Mesure d'atténuation |
|---|---|
| **Hallucination du LLM** | Prompt strict, filtrage par seuil de pertinence (Cross-Encoder), refus explicite. |
| **PDF illisibles / corrompus** | Pipeline d'ingestion renforcé par des heuristiques de nettoyage avancées. |
| **Saturation de l'API (Quota)** | Mise en cache des réponses fréquentes et bascule automatique sur un modèle secondaire. |
| **Erreur d'interprétation humaine**| Affichage de l'article source brut et disclaimer juridique permanent. |

## 12. Contraintes

- Interdiction d'utiliser des bases de connaissances externes au corpus fourni (Grounded only).
- Hébergement Cloud avec budget limité (utilisation des crédits gratuits ou d'infrastructures légères).

## 13. Évolutions Futures (Vision à long terme)

- **Intégration vocale** (Speech-to-Text) pour inclure les populations peu à l'aise avec la lecture/écriture.
- Support des langues nationales béninoises (Fon, Mina, etc.).
- Ajout de la jurisprudence et des lois modificatives en temps réel.
- Création d'une API pour intégration sur les sites gouvernementaux.
